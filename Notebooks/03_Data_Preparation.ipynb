```python
import pandas as pd
```

```python
sales = pd.read_csv('C:/Users/Samiksha/Desktop/Business_performance_Dashboard/Data/raw/sales.csv')
```

```python
sales_clean = sales.copy()
```

```python
product = pd.read_csv('C:/Users/Samiksha/Desktop/Business_performance_Dashboard/Data/raw/products.csv')
```

```python
product_clean = product.copy()
```

```python
stores = pd.read_csv('C:/Users/Samiksha/Desktop/Business_performance_Dashboard/Data/raw/stores.csv')
```

```python
stores_clean = stores.copy()
```

```python
sales_clean.head()
```

```python
sales_clean['Date'] = pd.to_datetime(
    sales_clean['Date'],
    format='%Y-%m-%d',
    errors = 'coerce'
)
```

```python
product_clean['Product_Cost']= pd.to_numeric(
    product_clean['Product_Cost'].str.replace('$','',regex=False),
    errors='coerce')
    
```

```python
product_clean['Product_Price']= pd.to_numeric(
    product_clean['Product_Price'].str.replace('$','',regex=False),
    errors='coerce')
    
```

```python
product_clean.head()
```

```python
stores_clean.head()
```

```python
print(sales_clean['Date'].isna().sum())
print(product_clean[['Product_Cost', 'Product_Price']].isna().sum())
print(stores_clean['Store_Open_Date'].isna().sum())
```

```python
sales_clean['Date'].dtypes
```

```python
sales_product=pd.merge(
    sales_clean,
    product_clean,
    on='Product_ID',
    how='left',
    validate='many_to_one')
```

```python
sales_final=pd.merge(
    sales_product,
    stores_clean,
    on='Store_ID',
    how='left',
    validate='many_to_one')
```

```python
sales_final['Revenue'] = sales_final['Units'] * sales_final['Product_Price']
```

```python
sales_final['COGS'] = sales_final['Units'] * sales_final['Product_Cost']
```

```python
sales_final['Gross_Profit'] = sales_final['Revenue'] - sales_final['COGS']
```

```python
sales_final[['Revenue', 'COGS', 'Gross_Profit']].isna().sum()
```

```python
import numpy as np

np.isclose(
    sales_final['Gross_Profit'],
    sales_final['Revenue'] - sales_final['COGS']
).all()
```
